```python
# standardizing and encode null detection flags as explicit yes or no 

flag_cols = ['Malware Indicators','Alerts/Warnings','Firewall Logs','IDS/IPS Alerts','Proxy Information']
for col in flag_cols:
    df[col]=df[col].notna().map({True:'Yes',False:'No'})

df['proxy_used'] =  df['Proxy Information'].notna().map({True:'Yes',False:'No'})

    
```

```python
df
```

```python
df[['Malware Indicators','Alerts/Warnings','Firewall Logs','IDS/IPS Alerts','Proxy Information']].head(10)
```

```python
df['Timestamp'] = pd.to_datetime(df['Timestamp'])    #converting the data type of the timestamp to datetime form string
```

```python
# parsing the timestamp data into seperate hour,date,day of week,year month for further timer analysis
```

```python
df['date'] = df['Timestamp'].dt.date
```

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
df['hour'] = df['Timestamp'].dt.hour
```

```python
df['day_of_week'] = df['Timestamp'].dt.day_name()
df['year_month'] = df['Timestamp'].dt.to_period('M')
```

```python
df['day_of_week'] = df['day_of_week'].astype('string')
```

```python
df['year_month']
```

```python
df['hour'] = df['hour'].astype(int)
```

```python
df.columns
```

```python
df[['Source Port', 'Destination Port']]
```

```python
# validating the source destination ports and flag well know ports (0-1023) vs registered/dynamic ports

df['source_port_category'] = np.where(
    df['Source Port']<=1023,
    'Well-Known',
     'Registered/Dynamic' 
)
df['destination_port_category'] = np.where(
    df['Destination Port']<=1023,
    'Well-Known',
     'Registered/Dynamic' 
)
```

```python
df
```

```python
# making a detection coverage score
flag_cols = ['Malware Indicators','Firewall Logs','IDS/IPS Alerts','Alerts/Warnings']
df['detection coverage score'] = (df[flag_cols]=='Yes').sum(axis=1)
```

```python
df['detection coverage score']
```

```python
df[['Action Taken','Severity Level']]
```

```python
# creating a response alignment flag
df['Response Alignment'] = np.where(
                           (df['Action Taken']!='Blocked') & (df['Severity Level']=='High'),
                           1,
                           0)
```

```python
# now comes the part of splitting the gepo location daa into different city region
df['Geo-location Data'].head(10)
```

```python
df['City'] = df['Geo-location Data'].str.split(',').str[0]
```

```python
df['City']
```

```python
df['State'] = df['Geo-location Data'].str.split(',').str[1].str.strip()
```

```python
df[['City','State']]
```

```python
# converting the cleaned dataset into csv format
df.to_csv('cleaned_cybersecurity_attacks.csv',index=False)
```
